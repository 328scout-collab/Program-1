## Statement of computational problem
Problem 2: Determine whether a polygon is convex.

Input: a sequence〈p1, p1, . . . , pn〉 of n ≥ 3 points. Each index represents a pair of (x,y) coordinates. Consective points are connected by line segments. The last point is connected to the first one.
Output: True is the polygon is convex; false if it is not.

## Decision Rule Description
Decision rule (exterior angles):
    Walk around the polygon. At each vertex we "turn" by some exterior angle.
    If we add up the SIZE of every turn (ignoring left/right), a convex
    polygon turns exactly 360 degrees in total. A concave polygon has at
    least one turn in the opposite direction, so it must over-turn to get
    back to where it started, and the total ends up greater than 360.

## pseudocode

    isConvex(points)

    n = number_of_points
    total_turn = 0

    for i = 0 to n - 1
        a = previous_point
        b = current_point
        c = next_point

       angle = exterior_angle(a, b, c)
       total_turn = total_turn + |angle|

    return !(total_turn > 360)

## Algorithm Justification
**Loop Invariant:** At the beginning of each iteration of the loop, total_turn contains the sum of the absolute exterior angles for all vertices that have already been processed.

**Initialization:** Before the first iteration, since no vertices have been processed, total_turns is initialized to be 0. Making the Loop invariant true at the start of the loop.

**Maintenance:** During each iteration of the loop, the algorithm selects the previous vertex "a", current vertex "b" and next vertex "c". It then calculates the exterior turning angle at b and adds the absolute value of this angle to total_turn, maintaining the loop invariant.

**Termination:** The loop executes once per vertice. When the loop terminates, every vertex has been processed so by the loop invariant total_turn is the sum of all the exterior angles. Based on our decision rule, a convex polygon has a total absolute exterior turn of 360 degrees. If the polygon is concave at least one of the turn would be on the opposite direction to the other turns making the absolute sum of the angles greater than 360 degrees. Thus, if total_turns is greater than 360 degrees, then the polygon is concave, otherwise it is convex.

## Implementation

```python
import math


def exterior_angle(a, b, c):
    """
    Return the exterior (turning) angle in degrees at point b,
    when walking from a -> b -> c.

    """
    in_x = b[0] - a[0]
    in_y = b[1] - a[1]

    out_x = c[0] - b[0]
    out_y = c[1] - b[1]

    cross = in_x * out_y - in_y * out_x
    dot = in_x * out_x + in_y * out_y

    return math.degrees(math.atan2(cross, dot))


def is_convex(points):
    """
    Return True if the polygon described by points is convex, otherwise False.

    points is a list of (x, y) coordinates, in order around the polygon.
    """
    n = len(points)
    total_turn = 0

    for i in range(n):
        a = points[i - 1]          # previous point (wraps to the last point)
        b = points[i]              # current point
        c = points[(i + 1) % n]    # next point (wraps to the first point)

        total_turn += abs(exterior_angle(a, b, c))

    return abs(total_turn - 360) < 1e-6
```

## Test Cases

```python
"""
Test cases for is_convex.  Run with:  python test_convex.py
"""


# (name, points, expected answer)
test_cases = [
    # ---------- Convex shapes ----------
    ("Triangle",                    [(0, 0), (4, 0), (2, 3)],                       True),
    ("Square (counter-clockwise)",  [(0, 0), (2, 0), (2, 2), (0, 2)],               True),
    ("Square (clockwise)",          [(0, 0), (0, 2), (2, 2), (2, 0)],               True),
    ("Rectangle",                   [(0, 0), (5, 0), (5, 1), (0, 1)],               True),
    ("Regular pentagon",            [(0, 2), (-1.9, 0.62), (-1.18, -1.62),
                                     (1.18, -1.62), (1.9, 0.62)],                   True),
    ("Hexagon",                     [(1, 0), (3, 0), (4, 2), (3, 4), (1, 4), (0, 2)], True),
    ("Negative coordinates",        [(-5, -5), (-1, -5), (-1, -1), (-5, -1)],       True),
    ("Very thin triangle",          [(0, 0), (100, 0), (50, 0.01)],                 True),
    ("Square with point mid-edge",  [(0, 0), (1, 0), (2, 0), (2, 2), (0, 2)],       True),

    # ---------- Concave shapes ----------
    ("Arrowhead / dart",            [(0, 0), (2, 1), (4, 0), (2, 4)],               False),
    ("L-shape",                     [(0, 0), (2, 0), (2, 1), (1, 1), (1, 2), (0, 2)], False),
    ("L-shape (clockwise)",         [(0, 2), (1, 2), (1, 1), (2, 1), (2, 0), (0, 0)], False),
    ("Star",                        [(0, 3), (1, 1), (3, 1), (1.5, -0.5), (2, -3),
                                     (0, -1.5), (-2, -3), (-1.5, -0.5), (-3, 1), (-1, 1)], False),
    ("Square with one dent",        [(0, 0), (4, 0), (4, 4), (2, 3), (0, 4)],       False),
    ("U-shape",                     [(0, 0), (3, 0), (3, 3), (2, 3), (2, 1),
                                     (1, 1), (1, 3), (0, 3)],                       False),
    ("Barely concave",              [(0, 0), (4, 0), (4, 4), (2, 3.99), (0, 4)],    False),
]

passed = 0
print(f"{'Shape':<30} {'Expected':<10} {'Got':<10} Result")
print("-" * 60)

for name, points, expected in test_cases:
    result = is_convex(points)
    status = "PASS" if result == expected else "FAIL"
    if result == expected:
        passed += 1
    print(f"{name:<30} {str(expected):<10} {str(result):<10} {status}")

print("-" * 60)
print(f"{passed}/{len(test_cases)} tests passed")
```

## Asymptotic Analysis

exterior_angle(a, b, c):
O(1) Time
A fixed number of operations occur in the function regardless of input.

is_convex(points):
O(n) Time
The for loop will execute n times, based on the input of n points. Inside each iteration of the loop all operations are constant time O(1). Therefore the time complexity of the function is n x O(1) = O(n)

Overall the solution has O(n) time complexity in the worst case scenario.

## Benchmarks

```python
"""
Test Benchmark Time
"""
import time

def generate_regular_polygon(num_vertices):
    """Generate points for a regular polygon with n vertices."""
    points = []
    for k in range(num_vertices):
        angle = 2 * math.pi * k / num_vertices
        points.append((math.cos(angle), math.sin(angle)))
    return points


print("\n" + "=" * 60)
print("BENCHMARK RESULTS")
print("=" * 60)
print(f"{'Number of Vertices (n)':<25} {'Time (seconds)':<20} {'Time per Vertex (ns)':<20}")
print("-" * 60)

sizes = [10_000, 50_000, 100_000, 500_000, 1_000_000]

for n in sizes:
    poly = generate_regular_polygon(n)

    # Measure execution time over multiple iterations for accuracy
    iterations = 5 if n <= 100_000 else 1
    start_time = time.perf_counter()
    for _ in range(iterations):
        _ = is_convex(poly)
    end_time = time.perf_counter()

    avg_time = (end_time - start_time) / iterations
    time_per_vertex_ns = (avg_time / n) * 1e9

    print(f"{n:<25,d} {avg_time:<20.6f} {time_per_vertex_ns:<20.2f}")
```
